```python
human = ground_clean.rename(columns={
    "overall": "overall_human",
    "food": "food_human",
    "service": "service_human",
    "price": "price_human",
    "environment": "environment_human"
})

local = ollama_clean.rename(columns={
    "overall": "overall_local",
    "food": "food_local",
    "service": "service_local",
    "price": "price_local",
    "environment": "environment_local"
})

cloud = groq_all_fixed.rename(columns={
    "overall": "overall_cloud",
    "food": "food_cloud",
    "service": "service_cloud",
    "price": "price_cloud",
    "environment": "environment_cloud"
})

print(human.shape, local.shape, cloud.shape)
```

```python
merged = human.merge(
    local,
    on="review_id",
    how="inner"
).merge(
    cloud,
    on="review_id",
    how="inner"
)

print(merged.shape)
print("Unique review IDs:", merged["review_id"].nunique())
```

```python
aspects = ["overall", "food", "service", "price", "environment"]

mae_results = []

for aspect in aspects:
    human_col = f"{aspect}_human"
    local_col = f"{aspect}_local"
    cloud_col = f"{aspect}_cloud"

    local_valid = merged[[human_col, local_col]].dropna()
    cloud_valid = merged[[human_col, cloud_col]].dropna()

    local_mae = (local_valid[human_col] - local_valid[local_col]).abs().mean()
    cloud_mae = (cloud_valid[human_col] - cloud_valid[cloud_col]).abs().mean()

    mae_results.append({
        "aspect": aspect,
        "local_mae": local_mae,
        "cloud_mae": cloud_mae
    })

mae_df = pd.DataFrame(mae_results)

mae_df
```

```python
std_results = []

for aspect in aspects:
    human_col = f"{aspect}_human"
    local_col = f"{aspect}_local"
    cloud_col = f"{aspect}_cloud"

    local_valid = merged[[human_col, local_col]].dropna()
    cloud_valid = merged[[human_col, cloud_col]].dropna()

    local_errors = (
        local_valid[human_col] - local_valid[local_col]
    ).abs()

    cloud_errors = (
        cloud_valid[human_col] - cloud_valid[cloud_col]
    ).abs()

    std_results.append({
        "aspect": aspect,
        "local_std": local_errors.std(),
        "cloud_std": cloud_errors.std()
    })

std_df = pd.DataFrame(std_results)

std_df
```

```python
recall_results = []

for aspect in aspects:
    human_col = f"{aspect}_human"
    local_col = f"{aspect}_local"
    cloud_col = f"{aspect}_cloud"

    relevant = merged[human_col].notna()

    local_recall = merged.loc[relevant, local_col].notna().mean()
    cloud_recall = merged.loc[relevant, cloud_col].notna().mean()

    recall_results.append({
        "aspect": aspect,
        "local_recall": local_recall,
        "cloud_recall": cloud_recall
    })

recall_df = pd.DataFrame(recall_results)

recall_df
```

```python
outlier_summary = []

for aspect in aspects:
    for model in ["local", "cloud"]:
        error_col = f"{aspect}_error_{model}"

        worst = merged.nlargest(5, error_col)[
            ["review_id", f"{aspect}_human", f"{aspect}_{model}", error_col]
        ].copy()

        worst["aspect"] = aspect
        worst["model"] = model

        outlier_summary.append(worst)

outliers = pd.concat(outlier_summary, ignore_index=True)

print(outliers.shape)
outliers.head(10)
```

```python
clean_outliers = []

for aspect in aspects:
    for model in ["local", "cloud"]:
        error_col = f"{aspect}_error_{model}"

        temp = merged.nlargest(5, error_col)[
            ["review_id", f"{aspect}_human", f"{aspect}_{model}", error_col]
        ].copy()

        temp = temp.rename(columns={
            f"{aspect}_human": "human_score",
            f"{aspect}_{model}": "model_score",
            error_col: "error"
        })

        temp["aspect"] = aspect
        temp["model"] = model

        clean_outliers.append(temp)

outliers_clean = pd.concat(clean_outliers, ignore_index=True)

print(outliers_clean.shape)
outliers_clean.head(10)
```

### Final Recomendation

The Cloud model performed better than the Local model across the main evaluation metrics.

For accuracy, the Cloud model achieved lower MAE values across all five aspects: overall, food, service, price, and environment. It was also more consistent, with lower standard deviation of errors across all aspects.

Recall showed the clearest difference. The Cloud model achieved very high coverage for overall, food, and service, and much stronger coverage than the Local model for price and environment. The Local model especially struggled with environment, where it frequently returned missing values.

The qualitative analysis showed that both models struggled with mixed or contradictory reviews. Some extreme outliers also revealed possible inconsistencies in the human Ground Truth labels, especially for environment and price.

From an operational perspective, the Cloud model was also much faster. Based on the experiment, processing 5,000 reviews was estimated to take about 18.6 minutes with Groq, compared with about 5.2 hours using the Local Ollama model on the tested hardware. The estimated Groq API cost for 5,000 reviews was only about $2.09.

Overall, the Cloud model is the recommended option for Velox Foods because it provided better accuracy, higher consistency, much better coverage, and significantly faster processing. The Local model remains useful when data privacy or avoiding external API costs is the main priority.
